# Importing the dependencies 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

# Data Collection and Processing 

```python
# Loading the dataset

breast_cancer_dataset = pd.read_csv("./sample_data/data.csv")
```

```python
print(breast_cancer_dataset)
```

```python
# Checking for more info on the data

breast_cancer_dataset.info()
```

```python
# Checking for null values

breast_cancer_dataset.isnull().sum()
```

```python
# Dropping the "unamed: 32" and the "id" column

breast_cancer_dataset.drop(columns=['Unnamed: 32', 'id'], axis=1, inplace=True)
```

```python
breast_cancer_dataset.head()
```

```python
# Feature encoding

#  The diagnosis has values; M & B.
# M - Malignant
# B - Benigh

# We will replace M with 0 and B with 1


breast_cancer_dataset['diagnosis'] = breast_cancer_dataset['diagnosis'].replace({
    "M": 0,
    "B": 1
})
```

```python
breast_cancer_dataset.head()
```

```python
# Getting the classification of the "diagnosis" column

breast_cancer_dataset["diagnosis"].value_counts()
```

```python
# Checking the statistical measures of the dataset

breast_cancer_dataset.describe()
```

```python
# Seperating the data into features and target

X = breast_cancer_dataset.drop(columns='diagnosis', axis=1)
Y = breast_cancer_dataset['diagnosis']
```

```python
X.head()
```

```python
Y.head()
```

#  Splitting the data into train and test data 


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

# Standardize the data 

```python
from sklearn.preprocessing import StandardScaler
```

```python
Scaler = StandardScaler()

X_train_std = Scaler.fit_transform(X_train)
X_test_std = Scaler.transform(X_test)
```

# Building the Neural Network 

```python
%pip install tensorflow
```

```python
# Importing tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
# Setting up the layers of my Neural Network

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20, activation='relu'),
    keras.layers.Dense(2, activation='sigmoid'),
])
```

```python
# Compiling the neural network

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy']
              )
```

```python
# Training the Neural Network

history = model.fit(X_train_std, Y_train, validation_split=0.1, epochs=10)
```

# visualizing the accuracy and the loss 


```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])


plt.title("Model Accuracy")
plt.ylabel('accuracy')

plt.xlabel('Epochs')
plt.legend(['training data', 'validation data '], loc = 'lower right')

```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])


plt.title("Model Loss")
plt.ylabel('Loss')

plt.xlabel('Epochs')
plt.legend(['training data', 'validation data '], loc = 'upper right')

```

# Accuracy of the model on test data 

```python
loss, accuracy = model.evaluate(X_test_std, Y_test)
```

```python
print(accuracy)
```
